```python
class Student:

    def __init__(self, name):
        self.name = name
        
    def greet(self):
        print(f"Hello {self.name} Welcome aboard!")

student1 = Student("Jaydeep")
Student2 = Student("Bhumi")
student3 = Student("Saish")
student4 = Student("Rahul")

student1.greet()
Student2.greet()
student3.greet()
student4.greet()
        
```

```python
class Student_Class:

    def __init__(self, name, age, Branch):
        self.name = name
        self.age = age
        self.Branch = Branch

    def Show_detail(self):
        print("\nName:",self.name)
        print("Age:",self.age)
        print("Branch:",self.Branch)

Student1 = Student_Class("Jaydeep", 21, "Extc")
Student2 = Student_Class("Bhumi", 21, "CSE")
Student3 = Student_Class("Shantanu", 24, "Extc")
Student4 = Student_Class("Sayali", 26, "Medical")
Student5 = Student_Class("Saish", 16, "Merchant Navy")

Student1.Show_detail()
Student2.Show_detail()
Student3.Show_detail()
Student4.Show_detail()
Student5.Show_detail()
```
